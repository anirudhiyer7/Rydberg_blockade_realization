```python
import matplotlib.pyplot as plt
import numpy as np
import pairinteraction as pi

ket1 = pi.KetAtom("Rb", n=79, l=2, j=2.5, m=2.5)
ket2 = pi.KetAtom("Rb", n=79, l=2, j=2.5, m=-2.5)
basis = pi.BasisAtom("Rb", n=(74, 84), l=(0, 4))
print(f"Number of single-atom basis states: {basis.number_of_states}")
system = pi.SystemAtom(basis)

pair_energy = ket1.get_energy(unit="MHz") + ket2.get_energy(unit="MHz")
min_energy, max_energy = pair_energy - 500, pair_energy + 500

pair_basis = pi.BasisPair(
    [system, system], energy=(min_energy, max_energy), energy_unit="MHz", m=(-5,5)
)
print(f"Number of two-atom basis states: {pair_basis.number_of_states}")

distances = np.linspace(0, 15, 200)
pair_systems = [
    pi.SystemPair(pair_basis).set_distance_vector([0, d, 10.2], unit="micrometer") for d in distances
]

# Diagonalize the systems in parallel
pi.diagonalize(pair_systems, diagonalizer="eigen")

eigenenergies = [system.get_eigenenergies(unit="MHz")-pair_energy for system in pair_systems]

eigenenergies = np.array(eigenenergies)
import plotly.graph_objects as go
```

```python
fig = go.Figure()

for j in range(1660):
    fig.add_trace(go.Scatter(
        x=distances,
        y=eigenenergies[:, j],
        mode="lines",
        opacity=1,
    ))
fig.update_layout(
    xaxis_title="Distance |y1-y2| (μm)",
    yaxis_title="U (MHz)",
    template = "plotly_dark",
    yaxis=dict(range=[-5, 0.5])
)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pairinteraction as pi

ket1 = pi.KetAtom("Rb", n=79, l=2, j=2.5, m=0.5)
ket2 = pi.KetAtom("Rb", n=79, l=2, j=2.5, m=0.5)
basis = pi.BasisAtom("Rb", n=(74, 84), l=(0, 4))
print(f"Number of single-atom basis states: {basis.number_of_states}")
system = pi.SystemAtom(basis)
system.set_magnetic_field([0, 0, 1.15e-3], unit="tesla")

system.diagonalize()                      # do this for B = 0 too
overlaps = system.get_eigenbasis().get_overlaps(ket1)
idx = np.argmax(overlaps)
E_r = system.get_eigenenergies(unit="MHz")[idx]
pair_energy = 2 * E_r
min_energy, max_energy = pair_energy - 500, pair_energy + 500

pair_basis = pi.BasisPair(
    [system, system], energy=(min_energy, max_energy), energy_unit="MHz"
)
print(f"Number of two-atom basis states: {pair_basis.number_of_states}")

distances = np.linspace(0, 15, 200)
pair_systems = [
    pi.SystemPair(pair_basis).set_distance_vector([0, d, 10.2], unit="micrometer") for d in distances
]

# Diagonalize the systems in parallel
pi.diagonalize(pair_systems, diagonalizer="eigen")

eigenenergies = [system.get_eigenenergies(unit="MHz")-pair_energy for system in pair_systems]
```

```python
eigenenergies = np.array(eigenenergies)
import plotly.graph_objects as go

fig = go.Figure()

for j in range(1542):
    fig.add_trace(go.Scatter(
        x=distances,
        y=eigenenergies[:, j],
        mode="lines",
        opacity=1,
    ))
fig.update_layout(
    xaxis_title="Distance |y1-y2| (μm)",
    yaxis_title="U (MHz)",
    template="plotly_dark",
    yaxis=dict(range=[-5, 5])
)
```
